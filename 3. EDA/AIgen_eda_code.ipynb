```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

folder = '/Users/alejandrogomez-paz/Desktop/UFC Project/2. data_cleaning/'

df1 = pd.read_csv(folder + 'fighters.csv')
df2 = pd.read_csv(folder + 'fighters_fights.csv')
df3 = pd.read_csv(folder + 'fights.csv')

def plot_by_round(df, col, max_round=5, sample=150, seed=0, q = 0.95, save_dir='round_plots'):
    """One subplot per fight length (1..max_round rounds).
    Each fighter-fight = one line of round-by-round `col` values."""

    d = df[df['round'] > 0].copy()                       # drop round-0 fight totals
    d['n_rounds'] = d.groupby('fight_id')['round'].transform('max')

    fig, axes = plt.subplots(1, max_round, figsize=(3.5*max_round, 4), sharey=False)

    for n, ax in zip(range(1, max_round + 1), axes):
        g = d[d['n_rounds'] == n]
        ids = g['fight_id'].drop_duplicates()
        keep = ids.sample(min(sample, len(ids)), random_state=seed)  # avoid overplotting 71k rows

        for _, traj in g[g['fight_id'].isin(keep)].groupby(['fight_id', 'fighter']):
            traj = traj.sort_values('round')
            ax.plot(traj['round'], traj[col], color='steelblue', alpha=0.08)

        m = g.groupby('round')[col].mean()               # mean trend on top
        ax.plot(m.index, m.values, color='crimson', marker='o', lw=2, label='mean')
        ax.set_title(f'{n}-rd fights (n={ids.nunique()})')
        ax.set_xticks(range(1, n + 1))
        ax.set_xlabel('round')
        ax.set_ylim(g[col].quantile(1 - q), g[col].quantile(q))

    axes[0].set_ylabel(col)
    axes[0].legend()
    fig.suptitle(f'{col} by round, split by fight length')

    os.makedirs(save_dir, exist_ok=True)
    fig.savefig(f'{save_dir}/{col}_by_round.jpg', dpi=150, bbox_inches='tight')
    plt.close(fig)   # suppresses display

# one column:
plot_by_round(df3, 'sig_str_landed')

# all numeric columns:
for c in df3.select_dtypes('number').columns.drop('round'):
    plot_by_round(df3, c)
```
